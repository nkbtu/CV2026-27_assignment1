***
#### Describe below how the matching, based on LOWE, works.

For each SIFT descriptor in the first image, it finds its two neares neighbors in the second image based on euclidean distance. Then it compares the distance of the best match to the distance of the second-best match. If the ratio of those is smaller than 0.8, the match is accepted, filtering out ambigous matches.
